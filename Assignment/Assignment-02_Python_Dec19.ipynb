### Math Tutor Using Random, Arithmetic Operators, and OOP

*Scenario:*
Create a `MathTutor` class that generates random math questions using `random` and `math operators`.

* Track correct and incorrect answers.
* Provide a score at the end.
* Handle invalid input using exception handling.

> Concepts: Random, Math, Exception Handling, Classes

```python
# Your Solution Here

import random

class MathTutor:
    def __init__(self, number_of_question = 5):
        self.q_num = number_of_question
        self.correct = 0
        self.incorrect = 0
        
    def generate_question(self):
        a = random.randint(1, 20)
        b = random.randint(1, 20)
        op = random.choice(['+', '-', '*'])
        answer = eval(f"{a} {op} {b}")
        return f"{a} {op} {b}", answer

    def ask(self):
        for i in range(1, self.number_of_question + 1):
            question, answer = self.generate_question()
            print(f"\nQ{i}: What is {question}?")
            try:
                user = int(input("Answer: "))
                print(f"Answer : ", user)
                if user == answer:
                    self.correct += 1
                else:
                    self.incorrect += 1
            except ValueError:
                print("Invalid input. Skipping.")
                self.incorrect += 1

    def score(self):
        print(f"\n--- Results ---")
        print(f"Correct:   {self.correct}/{self.number_of_question}")
        print(f"Incorrect: {self.incorrect}/{self.number_of_question}")
        print(f"Score:     {self.correct / self.number_of_question * 100:.0f}%")        
        
```
