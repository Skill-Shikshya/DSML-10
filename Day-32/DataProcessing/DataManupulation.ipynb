```python
import pandas as pd

titanic_df = pd.read_csv("update_titanic_data.csv")

titanic_df.head()
```

```python
# apply , map, aggregate

# def get_initials(name:str)-> str:
#     return name.split(",")[-1].split(".")[0].strip()

get_initials = lambda name : name.split(",")[-1].split(".")[0].strip()

titanic_df.Name.apply(get_initials)
```

```python
get_initials = lambda name : name.split(",")[-1].split(".")[0].strip()

titanic_df.Name.map(get_initials)
```

```python
get_max = lambda values: values.max()

titanic_df.Age.aggregate(["mean","sum", get_max])
```

### Handling Incorrect Data

```python
import numpy as np

data = pd.DataFrame({
    "id":[1,3,5,7,9],
    "name":["John","Jane","Jennifer","Johan","Jeff"],
    "age":[25,-32,19,42,"55"],
    "gender" : ["M","M","Z","F","F"],
    "income":[400, 700, 300, np.nan, 600]
})
```

```python
data
```

```python
data.dtypes
```

```python
data = data.astype({"age":"int16"})
data.age = abs(data.age)
```

```python
data[data.gender.isin(["M","F"])]
```
