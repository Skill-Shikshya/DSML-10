```python
import pandas as pd

titanic_df = pd.read_csv("titanic_data.csv")

titanic_df.head()
```

### Inserting or Deleting Columns and Rows

```python
# Dropping a column

titanic_df.drop(
    "PassengerId", axis=1, inplace=True
)

# titanic_df.drop(
#     ["PassengerId","Name"], axis=1, inplace=False
# )
```

```python
titanic_df.drop(
    [0,1,2], axis=0, inplace=False
)
```

```python
# intitals = []
# for name in titanic_df.Name:
#     initi = name.split(",")[-1].split(".")[0].strip()
#     intitals.append(initi)
    
# List Comprehensions
intitals = [ name.split(",")[-1].split(".")[0].strip() for name in titanic_df.Name]
```

```python
# Inserting or replacing Columns

titanic_df.insert(loc = 1, column="initials", value=intitals)
titanic_df.head()
```

```python
# Inserting or replacing Columns

titanic_df["new_initials"] = intitals
titanic_df.head()
```

```python
# Inserting or replacing Rows

titanic_df.loc[0, :] = [0, "Mr", 3, "Shrestha, Mr. Dinesh", "male", 56, 2, 0, "T33/4", 57.36 , "C234", "S", "Mr"]
titanic_df.head()
```
