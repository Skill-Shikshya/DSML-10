```python
def give_time(func):
    def wrapper(*args, **kwargs):
        import time
        start = time.time()
        res= func(*args, **kwargs)
        end = time.time()
        
        print(f"Function took {round(end-start, 3)} secs")
        
        return res
    return wrapper
```

```python
@give_time
def long_run(wait: int = 6):
    import time
    print("running")
    time.sleep(wait)
    print("complete")

    return wait
```

```python
long_run(4)
```
