## Reading Data From Files

- `pd.read_csv(file_path, sep=',')`: Reads CSV (Comma Separated Value) file and returns a Pandas DataFrame
    - *file_path*: Path to csv file
    - *sep*: Symbol separating different columns (default=',')

```python
titanic_df = pd.read_csv("titanic_data.csv", sep = ",", header = 0)
```

```python
titanic_df
```

- `pd.read_json(file_path, sep=',')`: Reads Json file and returns a Pandas DataFrame
    - *file_path*: Path to csv file
    - *sep*: Symbol separating different columns (default=',')

```python
employee = pd.read_json("generated.json", orient="records")
employee
```
