```python
import pandas as pd

titanic_data = pd.read_csv("titanic_data.csv")
```

```python
titanic_data.head()
```

## Condition based Filtering

```python
mask = titanic_data.Fare >= 100

fare_100 = titanic_data[mask]
```

```python
print(fare_100.shape)
fare_100.head()
```

```python
female_fare_100 = fare_100[fare_100.Sex == "female"]
```

```python
female_fare_100.shape
```

Out of 891 passengers from titanic , 53 were high paying passengers , which includes 65% of female and 35% of male passengers.

```python
# Female Survial Rate

survived_female = titanic_data[
    (titanic_data.Sex == "female") & (titanic_data.Survived == 1)
]
survived_female.shape
```

```python
# Survial rate according to Pclass

for i in [1,2,3]:
    temp = titanic_data[
        (titanic_data.Pclass == i) & (titanic_data.Survived == 1)
    ]
    print(f"Survived Passenger in Pclass {i} : {temp.shape[0]}")
```

```python
# isna or notna

null_cabin = titanic_data[titanic_data.Cabin.isna()]
null_cabin.shape
```

```python
# isna or notna

null_cabin = titanic_data[titanic_data.Cabin.notna()]
null_cabin.shape
```
